```python
# import my wandb project data
import wandb
api = wandb.Api()
runs = api.runs("ft_multiqa_tuneLR")
```

```python
eval_losses = [run.history()['eval/loss'] for run in runs]
```

```python
# remove nans
import pandas as pd
tmp_eval_losses = [[x[i] for i in range(len(x)) if not pd.isna(x[i])] for x in eval_losses]

# merge runs with the same name
from collections import defaultdict
d = defaultdict(list)
for name, loss in zip([run.name for run in runs], tmp_eval_losses):
    if len(loss) < 10:
        continue
    else:
        d[name] += loss
len(d)
```

```python
import matplotlib.pyplot as plt

gemma2bit_runs = [name for name in d.keys() if "gemma2bit" in name]
gemma2b_runs = [name for name in d.keys() if "gemma2b_" in name]

fig, axs = plt.subplots(1, 2, figsize=(12, 6))

for name in gemma2bit_runs:
    axs[0].plot(d[name], label=name.split("LR")[1])

for name in gemma2b_runs:
    axs[1].plot(d[name], label=name.split("LR")[1])

axs[0].set_ylabel('Eval Loss')
axs[1].set_ylabel('Eval Loss')

axs[0].set_xlabel('0.5k-steps')
axs[1].set_xlabel('0.5k-steps')

axs[0].set_title('Gemma2B-instruct')
axs[1].set_title('Gemma2B')

axs[0].legend(bbox_to_anchor=(1.05, 1), loc='upper left')
axs[1].legend(bbox_to_anchor=(1.05, 1), loc='upper left')

y_lim = [min(axs[0].get_ylim()[0], axs[1].get_ylim()[0]), max(axs[0].get_ylim()[1], axs[1].get_ylim()[1])]
axs[0].set_ylim(y_lim)
axs[1].set_ylim(y_lim)

plt.subplots_adjust(wspace=0.3)
plt.savefig('tmp/figs/gemma2b_instruct_vs_gemma2b.png')
plt.show()
```

```python
from modules.metrics import *
import json

attention_path = "experiments/bioasq12b/ll3_8b_instruct_FT_lora_attention_multiqa_splade_deberta_basicprompt_eval_splade_deberta_basicprompt/eval_dev_metrics_LLMeval_llama3.1:70b_out.json"
attention_path_2 = "experiments/bioasq12b/ll3_8b_instruct_FT_lora_attention_multiqa_splade_deberta_basicprompt_eval_splade_deberta_basicprompt/eval_dev_out.json"
mlp_path = "experiments/bioasq12b/ll3_8b_instruct_FT_lora_mlp_multiqa_splade_deberta_basicprompt_eval_splade_deberta_basicprompt/eval_dev_metrics_LLMeval_llama3.1:70b_out.json"
mlp_path_2 = "experiments/bioasq12b/ll3_8b_instruct_FT_lora_mlp_multiqa_splade_deberta_basicprompt_eval_splade_deberta_basicprompt/eval_dev_out.json"
with open(attention_path, 'r') as f:
    attention_data = [json.loads(line) for line in f]
with open(attention_path_2, 'r') as f:
    attention_data_2 = json.load(f)
with open(mlp_path, 'r') as f:
    mlp_data = [json.loads(line) for line in f]
with open(mlp_path_2, 'r') as f:
    mlp_data_2 = json.load(f)
```

```python
# merge the two attention data by question
attention_data = {x['question']: x for x in attention_data}
attention_data_2 = {x['question']: x for x in attention_data_2}
# for each key merge the dictionaries
for key in attention_data:
    attention_data[key].update(attention_data_2[key])

# merge the two mlp data by question
mlp_data = {x['question']: x for x in mlp_data}
mlp_data_2 = {x['question']: x for x in mlp_data_2}
# for each key merge the dictionaries
for key in mlp_data:
    mlp_data[key].update(mlp_data_2[key])
```

```python
attention_keys = set(attention_data.keys())
mlp_keys = set(mlp_data.keys())
intersection = attention_keys.intersection(mlp_keys)
assert len(intersection) == len(attention_keys) == len(mlp_keys)

# find questions where 'score' is 0 in attention but 1 in mlp
bad_attention_good_mlp = []
for q in intersection:
    if attention_data[q]['score'] == 0 and mlp_data[q]['score'] == 1:
        bad_attention_good_mlp.append(q)

# find questions where 'score' is 1 in attention but 0 in mlp
good_attention_bad_mlp = []
for q in intersection:
    if attention_data[q]['score'] == 1 and mlp_data[q]['score'] == 0:
        good_attention_bad_mlp.append(q)

decrease05_in_attention = []
for q in intersection:
    if attention_data[q]['score'] == mlp_data[q]['score'] - 0.5:
        decrease05_in_attention.append(q)

decrease05_in_mlp = []
for q in intersection:
    if mlp_data[q]['score'] == attention_data[q]['score'] - 0.5:
        decrease05_in_mlp.append(q)

print("0 attention, 1 MLP ==>", len(bad_attention_good_mlp), "\n1 attention, 0 MLP ==>", len(good_attention_bad_mlp))
print("decrease 0.5 in attention ==>", len(decrease05_in_attention), "\ndecrease 0.5 in MLP ==>", len(decrease05_in_mlp))
print("total questions", len(intersection))
```

```python
# 5 questions where attention is 0 but mlp is 1
for i,q in enumerate(bad_attention_good_mlp):
    assert attention_data[q]['instruction'] == mlp_data[q]['instruction']
    print(attention_data[q]["instruction"])
    print("Label: ", attention_data[q]["label"])
    print()
    print("LORA attention response:")
    print(attention_data[q]["response"])
    print(attention_data[q]["score"])
    print()
    print("LORA mlp response:")
    print(mlp_data[q]["response"])
    print(mlp_data[q]["score"])
    print()
    print()
```

- llmeval evaluates correctly on these cases (the metric is good)
- many yes no questions
- hypothesis: FT attention pays too much attention to all documents instead of prioritizing the top ones when the answer is there (reranking looses its usefulness)

```python
# 5 questions where attention is 0 but mlp is 1
for i,q in enumerate(good_attention_bad_mlp):
    if i == 2:
        assert attention_data[q]['instruction'] == mlp_data[q]['instruction']
        print(attention_data[q]["instruction"])
        print("Label: ", attention_data[q]["label"])
        print()
        print("LORA attention response:")
        print(attention_data[q]["response"])
        print(attention_data[q]["score"])
        print()
        print("LORA mlp response:")
        print(mlp_data[q]["response"])
        print(mlp_data[q]["score"])
        print()
        print()
```

```python
import os
import json
import matplotlib.pyplot as plt

dataset_dirs = {
    'experiments/bioasq12b': ['ll38binstruct_splade_deberta_bioasqprompt_epochs','experiments/bioasq12b/ll3_8b_splade_deberta_bioasqprompt_2'],
    'experiments/covidqa': ['ll38binstruct_splade_deberta_covidqaprompt_epochs','experiments/covidqa/llama3_8b_splade_deberta_covidqaprompt'],
    'experiments/fiqa': ['ll38binstruct_splade_deberta_fiqaprompt_epochs','experiments/fiqa/llama3_8b_splade_deberta_advprompt'],
    'experiments/paraphraserc': ['ll38binstruct_splade_deberta_paraphrasercprompt_epochs','experiments/paraphraserc/llama3_8b_splade_deberta_advprompt'],
    'experiments/robustqa_lifestyle': ['ll38binstruct_splade_deberta_lifestyleprompt_epochs','experiments/robustqa_lifestyle/llama3_8b_splade_deberta_lifestyleprompt'],
    'experiments/robustqa_recreation': ['ll38binstruct_splade_deberta_recreationprompt_epochs','experiments/robustqa_recreation/llama3_8b_splade_deberta_recreationprompt'],
    'experiments/robustqa_science': ['ll38binstruct_splade_deberta_scienceprompt_epochs','experiments/robustqa_science/llama3_8b_splade_deberta_scienceprompt'],
    'experiments/robustqa_technology': ['ll38binstruct_splade_deberta_technologyprompt_epochs','experiments/robustqa_technology/llama3_8b_splade_deberta_techprompt'],
    'experiments/robustqa_writing': ['ll38binstruct_splade_deberta_writingprompt_epochs','experiments/robustqa_writing/llama3_8b_splade_deberta_writingprompt'],
    'experiments/searchqa': ['ll38binstruct_splade_deberta_searchqaprompt_epochs','experiments/searchqa/llama3_8b_splade_deberta_techprompt'],
    'experiments/syllabusQA': ['ll38binstruct_splade_deberta_syllabusqaprompt_epochs','experiments/syllabusQA/llama3_8b_splade_deberta_syllqaprompt'],
    'experiments/techqa': ['ll38binstruct_splade_deberta_techqaprompt_epochs','experiments/techqa/llama3_8b_splade_deberta_techprompt'],
}

recall_along_epochs = {dataset_dir.split('/')[-1]: dict() for dataset_dir in dataset_dirs}
llmeval_along_epochs = {dataset_dir.split('/')[-1]: dict() for dataset_dir in dataset_dirs}

for dataset_dir,(base_dir,baseline_dir) in dataset_dirs.items():
    base_dir = os.path.join(dataset_dir, base_dir)
    # iterate folders in directory, find eval_dev_out.json, extract recall
    for folder in os.listdir(base_dir):
        with open(os.path.join(base_dir, folder, "eval_dev_metrics.json"), 'r') as f:
            data = json.load(f)
            epoch = folder.split("_")[-1].split('epoch')[0]
            epoch = float(epoch) if len(epoch) == 1 else float(f"{epoch[0]}.{epoch[1]}")
            recall_along_epochs[dataset_dir.split('/')[-1]][epoch] = data['Recall']
            if 'LLMeval_llama3.1:70b' in data:
                llmeval_along_epochs[dataset_dir.split('/')[-1]][epoch] = data['LLMeval_llama3.1:70b']
            else:
                llmeval_along_epochs[dataset_dir.split('/')[-1]][epoch] = None
    # add 0 epoch score
    with open(os.path.join(baseline_dir, "eval_dev_metrics.json"), 'r') as f:
        data = json.load(f)
        recall_along_epochs[dataset_dir.split('/')[-1]][0] = data['Recall']
        if 'LLMeval_llama3.1:70b' in data:
            llmeval_along_epochs[dataset_dir.split('/')[-1]][0] = float(data['LLMeval_llama3.1:70b'])

# sort along epochs
for dataset in recall_along_epochs:
    recall_along_epochs[dataset] = dict(sorted(recall_along_epochs[dataset].items()))
    llmeval_along_epochs[dataset] = dict(sorted(llmeval_along_epochs[dataset].items()))


# plot recall, llmeval on two plots
fig, axs = plt.subplots(1, 2, figsize=(15, 10))
colors = ['blue', 'red', 'green', 'orange', 'purple', 'brown', 'pink', 'gray', 'olive', 'cyan', 'magenta', 'yellow', 'black']


for i,(dataset,recalls) in enumerate(recall_along_epochs.items()):
    axs[0].plot(list(recalls.keys()), list(recalls.values()), label=dataset, color=colors[i])
axs[0].set_title("Recall along epochs")
axs[0].set_xlabel("Epoch")
axs[0].set_ylabel("Recall")

for i,(dataset,llmevals) in enumerate(llmeval_along_epochs.items()):
    axs[1].plot(list(llmevals.keys()), list(llmevals.values()), label=dataset, color=colors[i])
axs[1].set_title("LLMeval along epochs")
axs[1].set_xlabel("Epoch")
axs[1].set_ylabel("LLMeval")

axs[0].grid()
axs[1].grid()
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper center')
plt.savefig('tmp/figs/metrics_llamaFT_along_epochs.png')
plt.show()
```
